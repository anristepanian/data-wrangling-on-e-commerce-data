```python
# Sum of Unique Values by Column
for i in dt.columns:
    print(dt[i].value_counts())
```

```python
# Character Patterns
print(dt["Invoice"].str.replace(r"\d", "9", regex=True)
        .str.replace(r"[A-Za-z]", "A", regex=True).value_counts())
```

```python
print(dt["StockCode"].str.replace(r"\d", "9", regex=True)
        .str.replace(r"[A-Za-z]", "A", regex=True).value_counts())
```

```python
print(f'Number of rows where quantities are lower than 0: {dt["Quantity"][dt["Quantity"]<0].count()}')
print(f'Number of rows where quantities are equal to 0: {dt["Quantity"][dt["Quantity"]==0].count()}')
```

```python
print(f'Number of rows where price is lower than 0: {dt["Price"][dt["Price"]<0].count()}')
print(f'Number of rows where price equals to 0: {dt["Price"][dt["Price"]==0].count()}')
```

```python
# Invoice prefixes
dt.loc[dt["Invoice"].str.match(r"^[A-Za-z]"), "Invoice"].str[0].value_counts()
```

```python
# Relationship between invoices that start with "C" and negative quantities
c = dt["Invoice"].str.startswith("C")
print(pd.crosstab(c, dt["Quantity"] < 0))
```

```python
# Relationship between invoices that start with "A" and negative quantities
a = dt["Invoice"].str.startswith("A")
print(pd.crosstab(a, dt["Quantity"] < 0))
```

```python
# Negative Price Rows
dt[dt["Price"] < 0]
```

```python
# Zero Price Rows
dt[dt["Price"] == 0].head(10)
```

```python
# How many rows contain at least one lowercase letter
dt[dt["StockCode"].str.contains(r"[a-z]")]["StockCode"].nunique()
```

```python
# The single C invoice with non-negative quantity
dt[c & (dt["Quantity"] >= 0)]
```

```python
# The A invoices with non-negative quantity
dt[a & (dt["Quantity"] >= 0)]
```

# Guck mal Unter

```python
# Non-C rows with negative quantity
neg_noc = dt[~c & (dt["Quantity"] < 0)]
print(f'Proportion of missing description for quantities lower than 0 and non-C invoices: {(neg_noc["Description"].isna().mean())*100:.2f}%\n\n'
      f'Proportion of zero or negative prices for quantities lower than 0 and non-C invoices: '
      f'{((neg_noc["Price"] <= 0).mean())*100:.0f}%'
      f'\n\nProportion of missing Customer IDs for quantities lower than 0 and non-C invoices:'
      f' {(neg_noc["Customer ID"].isna().mean())*100:.0f}%')
```

```python
# Non-a rows with negative quantity
neg_noa = dt[~a & (dt["Price"] <= 0)]
print(f'Proportion of missing description for prices lower than or equal to 0 and non-A invoices: '
      f'{(neg_noa["Description"].isna().mean())*100:.2f}%\n\n'
      f'Proportion of negative quantities for prices lower than or equal to 0 and non-A invoices: '
      f'{((neg_noa["Quantity"] < 0).mean())*100:.2f}%'
      f'\n\nProportion of missing Customer IDs for prices lower than or equal to 0 and non-C invoices:'
      f' {(neg_noa["Customer ID"].isna().mean())*100:.2f}%')
```
